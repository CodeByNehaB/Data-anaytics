#### convert data coulmn to real dates

```python
date_cols =['order_purchase_timestamp', 'order_approved_at','order_delivered_carrier_date','order_delivered_customer_date','order_estimated_delivery_date']

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col],format='%d-%m-%Y %H:%M')

orders.dtypes
```

#### validate the order timeline

```python
print("Approved before purchase:", (orders['order_approved_at'] < orders['order_purchase_timestamp']).sum())
print("Carrier before approved:", (orders['order_delivered_carrier_date'] < orders['order_approved_at']).sum())
print("Delivered before carrier:", (orders['order_delivered_customer_date'] < orders['order_delivered_carrier_date']).sum())
print("Delivered before purchase:", (orders['order_estimated_delivery_date'] <orders['order_delivered_customer_date']).sum())
```
