### STEP 5 — Player Name Consistency

```python
## Check whether the same real player is ever recorded under more than one name spelling across the batter, bowler, and non_striker columns

import re
from collections import defaultdict

players = sorted(set(df['batter'].unique()) | set(df['bowler'].unique()))
print("Total Unique player names:", len(players))


#whitespace / case check

s = pd.Series(players)
print("ROW Uniques:", s.nunique(), "| stripped+lowercased uniques:", s.str.strip().str.lower().nunique())


# Do any names collapse to the same thing once dots and spaces are removed?
def signature(name):
    return re.sub(r'[.\s]', '', name).lower()

sig_map = defaultdict(list)
for p in players:
    sig_map[signature(p)].append(p)

dupes = {sig: names for sig, names in sig_map.items() if len(names) > 1}
print("Formatting duplicates (same name, different dots/spacing):", len(dupes))
for names in dupes.values():
    print(" ", names)

# Names with no space at all
single_token = sorted(p for p in players if ' ' not in p.strip())
print("\nSingle-token names:", single_token)

# Surnames shared by 4+ different name-strings (context, not automatically errors)
by_surname = defaultdict(list)
for p in players:
    parts = p.strip().split()
    if len(parts) >= 2:
        by_surname[parts[-1]].append(p)

common_surnames = {k: v for k, v in by_surname.items() if len(v) >= 4}
print(f"\nSurnames shared by 4+ distinct names: {len(common_surnames)}")
for k, v in common_surnames.items():
    print(f"  {k} ({len(v)} names)")
```

### STEP 11 — Standardize the city Column

```python
# Bangalore -> Bengaluru (current name; also resolves the M Chinnaswamy split automatically)
df['city_clean'] = df['city'].replace({"Bangalore": "Bengaluru"})

# 'Unknown' -> the venue's own real city (the correct value already exists in the data)
df.loc[(df['venue_clean'] == "Dubai International Cricket Stadium") & (df['city_clean'] == 'Unknown'), 'city_clean'] = "Dubai"
df.loc[(df['venue_clean'] == "Sharjah Cricket Stadium") & (df['city_clean'] == 'Unknown'), 'city_clean'] = "Sharjah"

# The two remaining either-way calls, resolved by whichever label is more common for that venue
df.loc[df['venue_clean'] == "Dr DY Patil Sports Academy", 'city_clean'] = "Mumbai"
df.loc[df['venue_clean'] == "Maharaja Yadavindra Singh International Cricket Stadium, Mullanpur", 'city_clean'] = "Mohali"

print("Cities before:", df['city'].nunique())
print("Cities after:", df['city_clean'].nunique())
print("'Unknown' remaining:", (df['city_clean'] == 'Unknown').sum())

# Re-check: does every venue now map to exactly one city?
recheck = df.groupby('venue_clean')['city_clean'].nunique()
print("Venues still mapping to more than one city:", (recheck > 1).sum())
```

### STEP 12 — Resolve and Parse win_outcome

```python
import re
import numpy as np

def parse_win_outcome(value):
    if pd.isnull(value):
        return np.nan, np.nan
    match = re.match(r'(\d+)\s+(runs|wickets)', str(value))
    if match:
        return int(match.group(1)), match.group(2)
    return np.nan, np.nan

parsed = df['win_outcome'].apply(lambda x: pd.Series(parse_win_outcome(x), index=['win_margin', 'win_type']))
df['win_margin'] = parsed['win_margin']
df['win_type'] = parsed['win_type']
```

```python
# Label the 25 previously-missing matches with WHY, instead of a silent blank
super_over_matches = [392190, 419121, 598004, 598017, 729315, 829741, 1082625, 1175365,
                       1178426, 1216493, 1216512, 1216517, 1216547, 1254077, 1473469, 1529281]
no_result_matches = [501265, 829763, 1359519, 1473492, 1473495, 1527685, 829813, 1178424, 1473481]

df.loc[df['match_id'].isin(super_over_matches), 'win_type'] = 'Tied (Super Over)'
df.loc[df['match_id'].isin(no_result_matches), 'win_type'] = 'No Result'

print("Row-level win_type breakdown:")
print(df['win_type'].value_counts(dropna=False))
```

```python
print("\nMatch-level win_type breakdown (more meaningful than row counts):")
print(df.groupby('match_id')['win_type'].first().value_counts())

print("\nwin_margin range by type:")
print(df[df['win_type'].isin(['runs', 'wickets'])].groupby('win_type')['win_margin'].agg(['min', 'max', 'mean']))
```
