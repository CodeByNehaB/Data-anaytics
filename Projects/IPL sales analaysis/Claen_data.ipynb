## IPL DATA ANALYSIS

### clean the dataset

```python
#import required libarary

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
##load the cvs file with the help of pandas

df = pd.read_csv("IPL_Dataset_2008_to_2026.csv", low_memory=False)
df
```

```python
#check how many row and column of there
df.shape
```

```python
#chckeing the missing values

df.isnull().sum()
```

```python
#one column has missing values: win_outcome has 4,980 blank rows.
df[df['win_outcome'].isnull()][['match_id', 'match_won_by', 'win_outcome']].drop_duplicates()
```

```python
#
df['win_outcome'] = df['win_outcome'].fillna('No Result')
```

```python
#checking one more time  ater fixing

df.isnull().sum()
```

### cleaning the season names

```python
## cleaing the season name because some are in the text or some yr in the integer
df['season'].unique()
```

```python
#check the unique year in the data
df['year'].unique()
```

```python
df['year'].dtype
```

```python
#cleaning the team names issue because team name is changing 
df['batting_team'].unique()
```

```python
## a mapping dictionary if you see the this old name replace with new name 

team_name_map={
    'Delhi Daredevils':'Delhi Capitals',
    'Kings XI Punjab':'Punjab Kings',
    'Royal Challengers Bangalore':'Royal Challengers Bengaluru',
    'Rising Pune Supergiant':'Rising Pune Supergiants'
}

team_columns = ['batting_team', 'bowling_team', 'toss_winner', 'match_won_by']

for col in team_columns:
    df[col] = df[col].replace(team_name_map)
```

```python
#checking the unique count again
print(df['batting_team'].unique())
print(df['toss_winner'].unique())
```

### Cleaning the date columns

```python
df['date']=pd.to_datetime(df['date'], dayfirst=True)
```

```python
#checking the date type
df['date'].dtype
```

```python
#print top 5 date 
df['date'].head()
```

### Checking city and venue for inconsistencies
