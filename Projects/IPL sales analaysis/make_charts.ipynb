## Visualizing the dataset

#### Chart1 - Team win percentage


```python
import pandas as pd
import matplotlib.pyplot as plt

summary = pd.read_csv('team_win_summary.csv', index_col=0)
top_scorers = pd.read_csv('top_scores.csv', index_col=0).iloc[:,0]
```

```python
summary_sorted = summary.sort_values('win_pct', ascending=False)
plt.figure(figsize=(12,6))
plt.bar(summary_sorted.index, summary_sorted['win_pct'])
plt.title("IPL Team Win Percentage (2008-2026)")
plt.xlabel("Team")
plt.ylabel("Win percentage")
plt.xticks(rotation=75, ha='right')
plt.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
### Chart 2 top 10 run scorers

top_10 = top_scorers.head(10)


plt.figure(figsize=(10,6))
plt.bar(top_10.index, top_10.values, color='darkorange')
plt.title("Top 10 Run Scorers in IPL History")
plt.xlabel("Player")
plt.ylabel("Total Career runs")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
