## Built-In-function

##### Built in function are predefined helper that are loaded automatically into memory by the interpreter.there are total 70 built in function in python

**1** ABS() calculate the absolute value of a number

```python
x = abs(-8)
print(x)
```

```python
y =abs(-7.5)
print(y)
```

**2** Bin() - coverts an integer into its binary represenatation

```python
x = bin(44)
print(x)
```

```python
x= bin(13)
print(x)
```

**3** Byte() - creates an unchangeable sequence of raw binary data.Each item in a bytes object is an integer from 0 to 255

```python
x = bytes(4)
print(x)
```

```python
x = bytes(16)
print(x)
```

**4** chr() - it convert an integer into corresponding character based on the unicode standard

```python
x = chr(98)
print(x)
```

```python
x = chr(23)
print(x)
```

**5** complex()  - creates a complex number .in math complex numbers have a real part and an imaginary part

```python
x = complex(3, 5)
print(x)
```

```python
x = complex(2.5, 3.96)
print(x)
```

```python
z = complex(2)
print(z)
```

**6** Float() - convert an integer , number, string or boolean value into floating-point number

```python
x = float(7)
print(x)
```

```python
y = float("11")
print(y)
```

```python
y = (float(True))
print(y)
```

```python
z = float("inf")
print(z)
```

**7 int()** - convert number or text into integer

```python
x =int(4.9)
print(x)
```

```python
x = int('12')
print(x)
```

**8** str() - convert an object into its string format

```python
x = str(123)
print(x)
```

```python
age =10
x = str(age)
print("I am a "  +x  +" years old")
```

**9** Help()- function starts Python's built-in help system. It is your interactive manual for looking up how to use modules, classes, functions, and keywords directly from your code terminal.

```python
help(print)
```

```python
help(int)
```

**10** input() - it take input from the user

```python
name = input("enter your name")
print(name)
```
