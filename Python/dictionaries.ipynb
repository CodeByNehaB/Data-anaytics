## Dictionaries

###### Dictionaries are the unordered collection of item. They store data in key value pair and key must be unique and immutable

#### Create dict

```python
empty_dict= {}
print(type(empty_dict))
```

```python
empty_dict = dict()
print(empty_dict)
```

```python
students = {"name": "krish", 
            "age": 32,
            "grade": "A"}

print(students)
```

#Accessing dict element



```python
students = {"name": "krish", 
            "age": 32,
            "grade": "A"}

print(students)
```

```python
print(students['name'])
print(students['age'])
```

```python
#Accessing using get() method

print(students.get('grade'))
print(students.get('last_name'))  # no key availble
```

#### Modifying dict element

##### dic are mutable , so you can add, update and delete element

```python
print(students)
```

```python
students['age'] = 33  ##update the value for the key
print(students)

students['address'] ='India'   ### added the new and value
print(students)

```

```python
del students['name'] #  delete key and value pair

print(students)
```

#### Dict method

```python
keys = students.keys()
print(keys)

values = students.values()
print(values)

item = students.items()
print(item)
```

```python
## shallow copy 

student_copy = students
print(student_copy)
```

```python
students['name'] = "neha"
print(students)
```

```python
studnet_copy1 = students.copy() ## shallow copy 
print(studnet_copy1)
print(students)
```

```python
students['name'] = "sahil"
print(students)
print(studnet_copy1)
```

#### Iterating over dict

```python
# can can use loops to iterative

## iterating over leys

for keys in students.keys():
    print(keys)
```

```python
for values in students.values():
    print(values)
```

```python
## Iterate over key value pair

for key, value in students.items():
    print(f"{key}: {value}")
```

```python
#Nested dictionaries

student = {
    "student1":{"name":"neha", "age":25},
    "student2":{"name":"nikhil","age":24}

    
}

print(student)
```

```python
#Access nested dic element

print(student["student1"]["name"])
print(student["student1"]["age"])
```

```python
#Iterating over nested dict
for student_id, student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

#### Dictonaries comprehensive


```python
square = {x:x**2 for x in range(5)}
print(square)
```

```python
#condition dict comprehension

even_dict = {x:x**2 for x in range(10) if x%2 ==0}
print(even_dict)
```

#### Practice

```python
##use a dict to count the frequency the elemnt in the list

numbers = [1,2,2,3,3,3,4,4,4,4]

frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number] =1
print(frequency)
```

```python
## merge 2 dic into 1
dict1 ={"a":1,"b":2}
dict2 = {"c":3,"d":4}

merged_dict = {**dict1,**dict2}
print(merged_dict)
```
