# Loops

##### Loops are used to execute a block of code repeatedly until a condition is met or all items in a sequence are processed.

#### For Loop

###### When you want to repeat code a specific number of item or iterate through a sequence

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

```python
#start with 1, end with 10 and step it 1 means increase by 1
for i in range (1, 10, 1):
    print(i)
```

```python
for i in range(10, 1, -2):
    print(i)
```

```python
## String

str = "Neha Bawankar"

for i in str:
    print(i)
```

#### While Loop

###### The while loop continue to execute as long as the condition is true

```python
count =0

while count <5:
    print(count)
    count = count+1
```

```python
count =0

while count %2 ==0:
    print(count)
    count = count +1
```

#### Loop Control statements

##### Break  - the break statement exits the loop permaturely

```python


for i in range(10):
  if i == 5:
       break
  print(i)
```

##### Continue - The continue statement skips the current iteration and continue with the next

```python
for i in range(10):
    if i % 2==0:
        continue
    print(i)
```

##### Pass - the pass statement is a null operation , it does nothing

```python
for i in range(5):
    if i==3:
        pass
    print(i)
```

#### Nested Loop - loop inside a loop

```python
for i in range(3):
    for j in range(2):
        print(f"i: {i}, and j : {j}")
```
