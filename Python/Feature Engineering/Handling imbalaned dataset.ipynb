## Handling imbalanced dataset

1.up Sampling
2.Down sampling

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(123)
```

```python
## cretae a dataframe with two classes

n_samples = 1000
class_0_ration =0.9
n_class_0 = int(n_samples * class_0_ration)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

#### Up Sampling

```python
## cteate a dataframe with imbalanced dataset

class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'features_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'features_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
```

```python
df= pd.concat([class_0,class_1]).reset_index(drop=True)
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## upsampling

df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
```

```python
df_minority_upsampled=resample(df_minority, replace=True,
         n_samples=len(df_majority),
         random_state=42
        )
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_unsampled = pd.concat([df_majority, df_minority_upsampled])
```

```python
df_unsampled['target'].value_counts()
```

### Down Sampling

```python
## cteate a dataframe with imbalanced dataset


n_samples = 1000
class_0_ration =0.9
n_class_0 = int(n_samples * class_0_ration)
n_class_1 = n_samples - n_class_0


class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'features_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'features_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})


df= pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
print(df['target'].value_counts())
```
