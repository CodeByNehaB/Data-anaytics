### SET

##### Set are built in data type in puthon to store collection of unique element.They are unorder, unindexed and do not allow duplicate elements. 

##### sets are useful for membership test, eleminatiing duplicate enteries and perform mathematical operations.

#### Create a set

```python
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set =set()
print(type(my_empty_set))
```

#### Basic set operations

```python
## Adding and removing
#add  -Inserts a single element into the set.
my_set.add(7)
print(my_set)
```

```python
## removing the elemnt from the set

my_set.remove(4)
print(my_set)
```

```python
#remove--Removes a specific element. If the item does not exist, it raises a KeyError.

my_set.remove(10)
print(my_set)
```

```python
#discard -Removes a specific element. If the item does not exist, it does nothing (no error is thrown). This is the safest method.

my_set.discard(10)
print(my_set)
```

```python
#pop - Removes and returns an arbitrary (random) element from the set. Raises a KeyError if the set is empty.

remove_element = my_set.pop()
print(remove_element)
print(my_set)
```

```python
## clear = Removes all elements, leaving the set completely empty.

my_set.clear()
print(my_set)
```

#### Set memebership test

```python
my_set ={1,2,3,4,5,6}
print(3 in my_set)
```

#### Mathematical operations

```python
set1 ={1,2,3,4,5,6}
set2 = {7,8,9,4,5,6}

###union -Combines all unique elements from both sets.
#A | B
#syntax - A.union(B)
union_set = set1.union(set2)

print(union_set)
```

```python
#Intersaction - Finds only the elements that exist in both sets.
#A & B
# syntax -A.intersection(B)

intersect_set = set1.intersection(set2)
print(intersect_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
# difference -Finds elements in set A that are not present in set B.
#syntax - A.difference(B)

#A - B
set1 ={1,2,3,4,5,6}
set2 = {7,8,9,4,5,6}

diff_set = set1.difference(set2)
print(diff_set)
```

```python
##  Symmetric Difference-Finds elements that are in either of the sets, but not in both (the opposite of intersection).
#A ^ B
# syntax- A.symmetric_difference(B)
sym_set = set1.symmetric_difference(set2)
print(sym_set)
```
