```python
#Packing and unpacking tuple
#packing
packed =1, "hello", 3.24,
print(packed)

#unpacking
a,v,d = packed
print(a)
print(v)
print(d)
```

```python
#unpacking with the star

numbers =(1,2,3,4,5,6)
first, *middle, last=numbers

print(first)
print(middle)
print(last)
```
