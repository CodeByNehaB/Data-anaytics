
### Assignment 6: Linear Algebra

1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.
2. Create two NumPy arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.

```python
import numpy as np

# 1. Create a 3x3 NumPy array (Matrix)
matrix = np.array([[4, 2, 1],
                   [1, 3, 0],
                   [2, 1, 5]])

# 2. Compute the determinant
determinant = np.linalg.det(matrix)

# 3. Compute the multiplicative inverse
inverse = np.linalg.inv(matrix)

# 4. Compute the eigenvalues (and eigenvectors)
eigenvalues, eigenvectors = np.linalg.eig(matrix)

# Display results
print("Original 3x3 Matrix:")
print(matrix)
print(f"\nDeterminant: {determinant:.2f}")
print("\nInverse Matrix:")
print(inverse)
print("\nEigenvalues:")
print(eigenvalues)
```

```python
arr1 = np.array([[1,2,3],
                [4,5,6]])
print("Array1:", arr1)

arr2 = np.array([[1,2],
                 [3,4],
                 [5,6]])
print("Array2:", arr2)


result = arr1 @ arr2
print("Result:", result)
```

```python
dt = np.dtype([('name', 'U20'), ('age', 'i4'), ('weight', 'f4')])

# 2. Add some sample data
data = [
    ('Alice', 25, 55.5),
    ('Bob', 32, 78.2),
    ('Charlie', 19, 68.0),
    ('Diana', 28, 62.1)
]

structured_array = np.array(data, dtype=dt)
print("Original Structured Array:")
print(structured_array)

sorted_array = np.sort(structured_array, order='age')

print("\nSorted Array by Age:")
print(sorted_array)
```
