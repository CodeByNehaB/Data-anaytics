## Seaborn

Seaborn in python visualization libaray based on the matplotlib that provide a high-level interface for drawing attractive and informative statical graphics

#### Main Component of seaborn
**Function** - 1. Figure level  2. Axis level
**Main Classification** -
1. Relational plot   2. distribution plot  3. categories plot   4. regression plot
5. matrix plot        6. multiplot

**Relational Plot** -

Generally used for -see the statical relation between 2 or more variable - bivariate analysis

Figure-Level Function- sns.relplot()
Axes-Level Equivalents -scatterplot(), lineplot()

```python
import seaborn as sns
```

```python
# Basic ploting with seaborn

tips = sns.load_dataset('tips')
tips
```

```python
# create a scatter plot

import matplotlib.pyplot as plt

sns.scatterplot(x = 'total_bill', y='tip', data=tips, hue='sex', style='time')
plt.title("Scatter plot of total bill vs Tip")
plt.show()
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

```python
## Create line plot
# sns.lineplot(data, x, y)
import plotly.express as px

gap= px.data.gapminder()
temp= gap[gap['country']=='India']
temp
```

```python
sns.lineplot(data=temp, x='year', y='lifeExp')
```

```python
sns.relplot(data=temp, x='year', y='lifeExp', kind='line')
```

```python
temp= gap[gap['country'].isin(['India', 'Brazil', 'Germany'])]
temp
```

```python
sns.relplot(data=temp, x='year', y='lifeExp', kind='line', hue='country', style='continent', size='continent')
```

##### Facet plot - figure level function only work iwth relplot



```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='line', col='sex', row='day')
```

**Distribution Plot** -

Generally used for -used for find out the distribution of data- show the range of data

Figure-Level Function- sns.displot()                     
Axes-Level Equivalents -histplot(), kdeplot(), rugplot()

```python
#histrogram plot
#sns.histplot(data, x, bins, hue, element) - element is used for desgging

sns.histplot(data= tips, x='total_bill', bins=10, hue='sex', element='step')
```

```python
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
## categorical plots

# hist plot

sns.histplot(x='day', y='total_bill', data= tips)
plt.title("hist plot of total bill")
plt.show()
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x= 'age', kind='hist', element='step', hue='sex')
```

**Bivariate Histrogram** - A bivariate histrogram bins the data within rectangle that tile the plot and then shows the count of observation within each rectangle with the fill color



```python
sns.displot(data=tips, x='total_bill', y='tip', kind='hist')
```

Kdelot() - Rather than using discrete bin a KDE plot smooths the observation with a Gaussian kenral producing a continues density estimate
**sns.kdeplot(data, x, hue, fill)**

```python
sns.kdeplot(data=tips, x='total_bill')
```

```python
sns.kdeplot(data=tips, x='total_bill', hue='sex', fill=True)
```

**Bivariate Kdeplot** - A bivariate kdeplot smoothes the (x,y) observation with 2d Gaussian

```python
sns.kdeplot(data=tips, x='total_bill', y='tip')
```

```python
sns.rugplot(data=tips, x='total_bill')
```

```python
# box plot

sns.boxplot(x='day', y='total_bill', data=tips)
plt.title("box plot of the total bill by day")
plt.show()
```
