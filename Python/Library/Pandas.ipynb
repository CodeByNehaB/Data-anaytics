Pandad is a powerful data manipulation library in python, widely used for data anaysis and data cleaning. It provide two primary data structure -                                                                                               
**Series** - it is one dimensional array like object                                                                                                                                                                                           
**Dataframe** - it is a two dimensinal array like matrix(row and column)

```python
import pandas as pd
```

### Series

```python
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
```

```python
## cRetae series from dictionary elemnt
data = {'a':1,'b':2,'c':3,'d':4}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index= ['a','b','c']
series = pd.Series(data, index= index)
print(series)
```

### DataFrame

```python
## create a dataframe from a dictionary of list

data = {
    'Name':['neha','alice','siya'],
    'Age':[20,25,28],
    'City':['Bengalore', 'Newyork','Florida']

}

df= pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## create a data frame from a list of dictionaries

data = [
    {'name':'krish', 'age':32,'City':'Bengalore'},
     {'name':'Nisraga', 'age':25,'City':'Bengalore'},
 {'name':'Nayan', 'age':32,'City':'Nagpur'}
]

df = pd.DataFrame(data)
print(df)
```
