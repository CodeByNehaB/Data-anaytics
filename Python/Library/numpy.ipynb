### Cpy and View in numpy

a copy creates a brand new independent array, while a view is just a window pointing to the original array's data.

1. **Using .view()** (Shared Data)Since a view shares memory with the original array, modifying one will modify both.

```python
original = np.array([1, 2, 3, 4])
arr_view = original.view()

# Modify the view
arr_view[0] = 99

print("Modified View:", arr_view)  
print("Original Array:", original)
```

2. **Using .copy()** (Independent Data)A copy is a completely separate object in your computer's memory.

```python
original = np.array([1, 2, 3, 4])
arr_copy = original.copy()

# Modify the copy
arr_copy[0] = 99

print("Modified Copy:", arr_copy)  
print("Original Array:", original)
```
