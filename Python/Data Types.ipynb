## DATA TYPES IN PYTHON

```python
Data Types are a classification of the data which tell the compiler or interpreter how the programmer intends to use the data

They determined the type of operation that can be performed on the data, the values that the data can take and the amount of memory needed to store the data.
```

## Importance of Data Types in Programming


```python
Data types ensure that data is stored in an efficient way
They help in performing correct operation on data
Proper use of data types can prevent and bugs in the program
```

```python
## Integer Example

age = 25
type(age)
```

```python
##floating point datatypes

height = 5.4
print(height)
print(type(height))
```

```python
##String datatype example

name ="Krish"
print(name)
print(type(name))
```
