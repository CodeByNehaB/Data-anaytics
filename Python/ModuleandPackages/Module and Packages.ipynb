## Modules and Packages

modules and packages are tools that help you organize large amounts of code into smaller, reusable pieces.

**Modules** -A module is simply any file ending in .py that contains Python code (variables, functions, or classes).

**Packages** - A package is a directory (folder) that holds multiple modules grouped together. It allows you to organize your modules hierarchically using "dot notation" (like import folder.file).

```python
import math

math.sqrt(16)
```

```python
from math import sqrt, pi
print(sqrt(16))

print(sqrt(25))
print(pi)
```
