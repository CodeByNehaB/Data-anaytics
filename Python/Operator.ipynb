## Operators

 Operator are special symbol or Keyword that perform specific computations on variable and value(operands).
 Python language support some types -

 1) Arithmetic Operator
 2) Assignment operator
 3) Comparioson Operator
 4) Logical operator
 5) Identify operator
 6) Membership operator
 7) Bitwise operator
    

### Arithmetic operator

```python
## Arithmetic operator

a= 5
b =10

add_result = a+b  #Addition
sub_result= a-b   #Substraction
mult_result = a*b #Multiplication
div_result = a/b  #Division
floor_result = a//b #Floor division
mod_result = a%b  #Modulus
exp_result = a** b  #Exponentiation

print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(floor_result)
print(mod_result)
print(exp_result)
```

### Logical Operators

```python
## And, Not , OR

x = True
y= True

result = x and y   #Both operand value should be true
print(result)
```

```python
##OR 
x = True
y= False
result = x or y # in Both any one operand value should be true
print(result)
```

```python
## Not
##It is reverse of boolean operator

x = True
y= False

result = not x
print(result)
```
