# Filepath

 a file path is the address that tells your script exactly where a file or folder lives on your computer's storage drive.To work with paths efficiently and prevent cros

```python
import os
cwd = os.getcwd()
print(f"Current work directory {cwd}")
```

```python
## create a new directory

new_directory = "package"
os.mkdir(new_directory)
print(f" Directory {new_directory} created")
```

```python
## Listing file and directory

items = os.listdir('.')
print(items)
```

```python
## joing paths
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd(),dir_name, file_name)
print(full_path)
```

```python
path = 'example.txt'
if os.path.exists(path):
    print(f" The path {path} exits")
else:
    print(f"The path {path} doesn't exits")
```

```python
import os

path = 'package'

if os.path.isfile(path):
    print(f" The path {path} is a file")
elif os.path.isdir(path):
    print(f"The path {path} is a directory")
else:
    print(f"The path {path} is neither a file nor a direcotry")
```
