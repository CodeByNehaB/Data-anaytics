## File Handling

 ###### File handling in python allow you to create , read, write and update and delete files direcly from your code.

###### 🔑 The Golden Rule: The with StatementIn Python, you should always use the with open() syntax. It automatically closes the file for you when the code block finishes, which protects your computer's memory and prevents data corruption.

```python
## Read a whole fiel

with open('example.txt', mode='r') as file:
    content = file.read()
    print(content)
```

```python
# read a file line by line

with open('example.txt', mode = 'r') as file:
    for line in file:
        print(line)
```

```python
# strpi - remove the line line space

with open('example.txt', mode = 'r') as file:
    for line in file:
        print(line.strip())
```

```python
# wrtiitng a file(overwrtting)

with open('example.txt', mode ='w') as file:
      file.write('hello world!\n')


```

```python
## write a file without overwritien

with open('example.txt', mode='a') as file:
    file.write("i added extra data\n")
    
```

```python
### writting a list of line to a file


lines = ['First line\n', 'second line\n', 'Thirs line\n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
## Read the content from a source file and write to a destination text file

with open('example.txt' , mode= 'r') as source_file:
    content = source_file.read()

with open('destination.txt', mode='w') as destination_file:
    destination_file.write(content)
```

```python
# read a text file and count the lines, wor and character

def count_text_file(file_path):
    with open(file_path, mode='r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count= sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, character = count_text_file(file_path)
print(f"Lines:{lines}, words:{words}, Character:{character}")
```
