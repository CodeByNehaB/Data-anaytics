## Exception Handling

Exception handling is a built-in safety net in Python. It prevents your program from crashing when an error happens, allowing you to catch the mistake, handle it gracefully, and keep your code running.

### Exception

Exception are events that discrupt the normal flow of program. They occur when an error is encountered during program execution.

```python
a=b
```

```python
# exception try, except block
#try: You place the code you think might fail here.
try: 
    a=b
#except: If an error happens inside the try block, Python jumps here instantly to run your backup safety plan
except:
    print("The variable has not been assigned")
```

```python
#A **NameError** happens when Python tries to use a variable, function, or library name that has not been created yet
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
#A **ZeroDivisionError** happens when your code tries to divide a number by zero

try:
    result =1/0
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter the denominator greater than 0")
```

```python
try:
    num = int(input("Enter a number"))
    result = 10/num
    print(f"Result:{result}")
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("You can not enter denominator greater than 0")
```

```python
try:
    num= int(input("Enter a number"))
    result = 1100/num
   
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("You can not enter denominator greater than 0")
#else: Runs only if the code inside the try block was completely successful with zero errors.
else:
    print(f"The result is {result}")
```

```python
try:
    num= int(input("Enter a number"))
    result = 110/num
   
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("You can't divide by zero")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")
finally:
    print("Execution complete")
```

```python
## FIle handling and exception handling

try:
    file = open('example.txt', 'r')
    content = file.read()
    print(content)
except FileNotFoundError:
    print("The file does not exists")
finally:
    if 'file' in locals() and not file.closed():
        file.close()
        print('file close')
```
