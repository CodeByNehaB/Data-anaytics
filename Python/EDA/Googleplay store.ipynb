```python
df_copy = df.copy()
```

```python
df_copy=df_copy.drop(df_copy.index[10472])
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
## conver Review Datattype to int

df_copy['Reviews']=df_copy['Reviews'].astype(int)
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size']=df_copy['Size'].str.replace('M', '000')
df_copy['Size']=df_copy['Size'].str.replace('k','')
df_copy['Size']=df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Size']
```

```python
df['Price'].unique()
```

```python
chars_to_remove=['+',',','$']
cols_to_clean=['Installs','Price']
for item in chars_to_remove:
    for cols in cols_to_clean:
        df_copy[cols]=df_copy[cols].str.replace(item,'')
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs']=df_copy['Installs'].astype('int')
```

```python
df_copy['Price']=df_copy['Price'].astype('float')
```

```python
df.info()
```

```python
## Handling last update features
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
df_copy['date']=df_copy['Last Updated'].dt.day
df_copy['Month']=df_copy['Last Updated'].dt.month
df_copy['Year']=df_copy['Last Updated'].dt.year
```

```python
df_copy.info()
```

```python
## save the cleanign dataset

df_copy.to_csv('google_clean.csv')
```

## explore data more

```python
numeric_features =[feature for feature in df_copy.columns if df_copy[feature].dtype != 'o']
categorical_features=[feature for feature in df_copy.columns if df_copy[feature].dtype != 'o']

## print columns
print('We have {} numerical features : {}'.format(len(numeric_features),numeric_features))
print('\nWe have {} categorical features"{}'.format(len(categorical_features),categorical_features))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the canvas grid layout
plt.figure(figsize=(15, 15))
plt.suptitle('Univariate Analysis of Numerical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.0)

# Loop through all numerical features and plot their distribution
for i in range(0, len(numeric_features)):
    plt.subplot(5, 3, i + 1)
    sns.kdeplot(x=df_copy[numeric_features[i]], fill=True, color='r')
    plt.xlabel(numeric_features[i])

plt.tight_layout()
plt.show()
```

## obervations
Rating and Year is left skewed while Review , size, Insatllls and price are right skewed

```python
plt.figure(figsize=(15, 15))
plt.suptitle('Univariate Analysis of Categorical fearutes', fontsize=20, fontweight='bold', alpha=0.8, y=1.0)
category =['Type', 'Content Rating']
# Loop through all numerical features and plot their distribution
for i in range(0, len(category)):
    plt.subplot(2, 2, i + 1)
    sns.countplot(x=df_copy[category[i]], palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)


plt.tight_layout()
plt.show()
```
