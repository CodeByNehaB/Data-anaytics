**Question 9** - Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

```python
def transpose_matrix(matrix):
    transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposed

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposed = transpose_matrix(matrix)
print("original matrix")

for row in matrix:
    print(row)

print("transposed matrxi:")
for row in transposed:
    print(row)
```

**Question 10** - Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

```python
def flattens_list(nested_list):
    flat_list = [item for sublist in nested_list for item in sublist]
    return flat_list

nested_list = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

flattened = flattens_list(nested_list)
print("original nested_list")
for row in nested_list:
    print(row)
print("flattern_list")

print(flattened)
```

**Question 13** Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists.

```python
def reverse_lst(input_lst):
    return input_lst[::-1]

original_list =[10,20,30,40,50]
print(f"original list:",original_list)

reversed_lst = reverse_lst(original_list)
print(f"reverse list:", reversed_lst)
```

**Question14** - Write a function that rotates a list by n positions. Print the original and rotated lists.

```python
def rotate_list(input_lts, n):
    if not input_lts:
        return input_lts

    n = n%len(input_lts)

    return input_lts[-n:] + input_lts[:-n]


ori_list =[1,2,3,4,5]
shift =3

rotated = rotate_list(ori_list, shift)
print(rotated)
    
```

**Question 15** - Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
def get_intersection(list1, list2):
    return list(set(list1).intersection(list2))


lst1 =[1,2,3,4,5]
lst2 =[2,4,4,1,5]

result = get_intersection(lst1, lst2)

print(result)
```
