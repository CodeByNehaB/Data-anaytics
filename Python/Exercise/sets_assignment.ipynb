**Question3** Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

```python
set1 = {1,2,3,4,5}
set2 = {2,4,5,8,10}
#uion
union_set = set1.union(set2)
print(union_set)

#intersection
inter_set = set1.intersection(set2)
print(inter_set)

#diffrefnce

diff_set = set1.difference(set2)
print(diff_set)

#symmetric difference

symdef = set1.symmetric_difference(set2)
print(symdef)
```
