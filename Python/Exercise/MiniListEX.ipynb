### 1: Manage a todo List

#### Create a to do list keep track of tasks


```python
#create a varibale 

to_do_list = ["buy groceries", "Clean the house", "Pay the bills"]

## Adding to task in todo list
to_do_list.append("Time to study")
to_do_list.append("going to college")

## removing a completed task
to_do_list.remove("Clean the house")
## checking if a task in the list or not

if "Pay the bills" in to_do_list:
    print("Don't forget the pay the utility bills")

print("To do list remaining")

for task in to_do_list:
    print(f" -{task}")
```

### Organizing student grades

#### Create a list to store and calculate average grade for student

```python
#organizing the student grades

grades = [85,79,94,90,88]

#adding a new grades

grades.append(75)

#calculating the average grade
average_grade = sum(grades)/len(grades)
print(f"Average Grade: {average_grade}")

#finding the highest and the lowest grade

highest_grade = max(grades)
lowest_grade = min(grades)

print(f"Highest grade:{highest_grade}")
print(f"lowest grade: {lowest_grade}")
```

### Managing a inventroy

#### use a list to manage inventroy items in store


```python
# Managing the inventory 

inventory = ["Apple", "Banana", "Cheery", "Grapes","Mnago"]

#Adding item in inventory

inventory.append("Watermelon")

#removing the item that out of stock

inventory.remove("Grapes")


#checking if an item is in stock

item = "orages"

if item in inventory:
    print(f"{item} are in stock")
else:
    print(f"{item} are not in stock")

print("Inventory list:")
for inv in inventory:
  print(inv)
```
