
## Simple calculator example

```python
num1 = float(input("Enter your first number: "))
num2 = float(input("Enter your second number: "))

sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1/num2

print("Sum: ", sum)
print("Difference: ", difference)
print("Product: ", product)
print("Quotient: ", quotient)
```
